```python

#KPI engineering
#KPI 1: Transfer Efficiency Ratio


df['Transfer Efficiency Ratio']=(
    df['Children transferred out of CBP custody']
    /
    df['Children in CBP custody']
)

df[['Transfer Efficiency Ratio']].head()

df['Transfer Efficiency Ratio'].mean()
print(f"Average Transfer Efficiency: {df['Transfer Efficiency Ratio'].mean()*100:.2f}%")
```

```python
#KPI 2- Discharge Effectiveness Index

df['Discharge Effectiveness']=(
    df['Children discharged from HHS Care']
    /
    df['Children in HHS Care']
)
discharge_effectiveness=df['Discharge Effectiveness'].mean() *100
print(f"Discharge Effectiveness Index: {discharge_effectiveness:.2f}%")
```

```python
#KPI 3 – Pipeline Throughput Rate

df['Pipeline Throughput Rate'] = (
    df['Children discharged from HHS Care']
    /
    df['Children apprehended and placed in CBP custody*']
)

pipeline_throughput = df['Pipeline Throughput Rate'].mean() * 100

print(f"Pipeline Throughput Rate: {pipeline_throughput:.2f}%")
```

```python
total_entries = df['Children apprehended and placed in CBP custody*'].sum()

total_exits = df['Children discharged from HHS Care'].sum()

pipeline_throughput = (total_exits / total_entries) * 100

print(f"Pipeline Throughput Rate: {pipeline_throughput:.2f}%")
```

```python
#KPI 4 – Backlog Accumulation Rate

df['Backlog Accumulation']=(
    df['Children apprehended and placed in CBP custody*']
    -
    df['Children discharged from HHS Care']
)

df['Backlog Accumulation'].describe()
```

```python
df['Backlog Accumulation'].describe()
print("Average Backlog:", df['Backlog Accumulation'].mean())
```

```python
#KPI 5 – Outcome Stability Score

outcome_stability = df['Children discharged from HHS Care'].std()

print(f"Outcome Stability Score (Std Dev): {outcome_stability:.2f}")
```

```python
# ==========================================
# 7. MONTHLY ANALYSIS
# ==========================================

monthly_data = (
    df
    .resample("ME")
    .sum(numeric_only=True)
)

monthly_data.head()
```

```python
# Monthly Apprehensions Trend

plt.figure(figsize=(12,5))

plt.plot(
    monthly_data.index,
    monthly_data['Children apprehended and placed in CBP custody*'],
    marker='o'
)

plt.title("Monthly Children Apprehended")
plt.xlabel("Month")
plt.ylabel("Children")
plt.grid(True)

plt.show()
```

```python
# Monthly discharge Trend
plt.figure(figsize=(12,5))

plt.plot(
    monthly_data.index,
    monthly_data['Children discharged from HHS Care'],
    marker='o'
)

plt.title("Monthly Children Discharged")
plt.xlabel("Month")
plt.ylabel("Children")
plt.grid(True)

plt.show()
```

```python
# ==========================================
# Monthly Apprehended vs Discharged
# ==========================================

plt.figure(figsize=(12,6))

plt.plot(
    monthly_data.index,
    monthly_data['Children apprehended and placed in CBP custody*'],
    marker='o',
    label='Apprehended'
)

plt.plot(
    monthly_data.index,
    monthly_data['Children discharged from HHS Care'],
    marker='s',
    label='Discharged'
)

plt.title("Monthly Apprehended vs Discharged")
plt.xlabel("Month")
plt.ylabel("Children")

plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
# KPI Values

transfer_efficiency = df['Transfer Efficiency Ratio'].mean() * 100

discharge_effectiveness = df['Discharge Effectiveness'].mean() * 100

pipeline_throughput = (
    df['Children discharged from HHS Care'].sum()
    /
    df['Children apprehended and placed in CBP custody*'].sum()
) * 100

outcome_stability = df['Children discharged from HHS Care'].std()

average_backlog = df['Backlog Accumulation'].mean()
```

```python
kpi_summary = {
    "Transfer Efficiency (%)": transfer_efficiency,
    "Discharge Effectiveness (%)": discharge_effectiveness,
    "Pipeline Throughput (%)": pipeline_throughput,
    "Outcome Stability": outcome_stability,
    "Average Backlog": average_backlog
}

```

```python
kpi_df = pd.DataFrame(
    kpi_summary.items(),
    columns=["KPI", "Value"]
)

kpi_df
```

```python
dashboard_kpis = pd.DataFrame({
    "KPI": [
        "Transfer Efficiency (%)",
        "Discharge Effectiveness (%)",
        "Pipeline Throughput (%)",
        "Outcome Stability",
        "Average Backlog"
    ],
    "Value": [
        transfer_efficiency,
        discharge_effectiveness,
        pipeline_throughput,
        outcome_stability,
        average_backlog
    ]
})

dashboard_kpis
```
